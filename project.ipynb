```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

df=pd.read_csv("House_Prediction.csv")
df.head()
```

```python
df.columns
df.shape
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
avg_price = df.groupby("bedrooms")["price"].mean()

plt.figure(figsize=(8,5))
plt.bar(avg_price.index, avg_price.values, color="skyblue")
plt.title("Average Price by Bedrooms")
plt.xlabel("Bedrooms")
plt.ylabel("Average Price")
plt.show()
```

```python
df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["sqft_living"], df["price"])
plt.xlabel("Living Area (sqft)")
plt.ylabel("Price")
plt.title("Living Area vs Price")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["price"], bins=20)
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.title("Price Distribution")
plt.show()
```

```python
avg_price = df.groupby("bedrooms")["price"].mean()

plt.figure(figsize=(8,5))
plt.bar(avg_price.index, avg_price.values, color="skyblue")
plt.title("Average Price by Bedrooms")
plt.xlabel("Bedrooms")
plt.ylabel("Average Price")
plt.show()
```

```python
bedroom_counts = df["bedrooms"].value_counts()
plt.figure(figsize=(8,8))
plt.pie(
    bedroom_counts,
    labels=bedroom_counts.index,
    autopct="%1.1f%%",
    startangle=90
)
plt.title("Distribution of Bedrooms")
plt.show()
```

```python
avg_price = df.groupby("floors")["price"].mean()
plt.figure(figsize=(8,5))
plt.plot(avg_price.index, avg_price.values, marker="o")
plt.title("Average Price by Floors")
plt.xlabel("Floors")
plt.ylabel("Average Price")
plt.grid(True)
plt.show()
```
